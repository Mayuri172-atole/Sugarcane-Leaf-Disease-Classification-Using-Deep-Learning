```python

# Step 4: Import all necessary packages
import os
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from PIL import Image
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.applications.resnet50 import preprocess_input, ResNet50
from tensorflow.keras.layers import GlobalAveragePooling2D, Dropout, Dense
from tensorflow.keras.models import Sequential
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau, ModelCheckpoint
from sklearn.metrics import confusion_matrix, classification_report
import cv2
```

```python
# Step 5: Dataset cleaning function
def clean_dataset(path, min_size=50):
    corrupted = []
    small = []
    for root, _, files in os.walk(path):
        for fname in files:
            if fname.lower().endswith(('png', 'jpg', 'jpeg')):
                fpath = os.path.join(root, fname)
                try:
                    with Image.open(fpath) as img:
                        img.verify()
                    with Image.open(fpath) as img2:
                        if img2.size[0] < min_size or img2.size[1] < min_size:
                            small.append(fpath)
                except (IOError, SyntaxError):
                    corrupted.append(fpath)
    for f in corrupted + small:
        os.remove(f)
    print(f"Removed {len(corrupted)} corrupted and {len(small)} small images")

DATASET_PATH = '/content/sugarcane_data'  # Check this path and adjust if needed
clean_dataset(DATASET_PATH)
```

```python

# Step 6: Custom augmentation functions for noise and blur
def add_noise(img_array):
    noise = np.random.normal(0, 25, img_array.shape).astype(np.uint8)
    return np.clip(img_array + noise, 0, 255).astype(np.uint8)

def add_blur(img_array):
    return cv2.GaussianBlur(img_array, (5,5), 0)

def custom_preprocessing(img):
    img_array = np.array(img)
    if np.random.rand() > 0.5:
        img_array = add_noise(img_array)
    if np.random.rand() > 0.5:
        img_array = add_blur(img_array)
    return Image.fromarray(img_array)
```

```python

# Step 7: Prepare ImageDataGenerator with augmentations
train_datagen = ImageDataGenerator(
    preprocessing_function=preprocess_input,
    rotation_range=40,
    width_shift_range=0.2,
    height_shift_range=0.2,
    shear_range=0.2,
    zoom_range=0.3,
    horizontal_flip=True,
    vertical_flip=True,
    fill_mode='nearest',
    validation_split=0.2
)

train_generator = train_datagen.flow_from_directory(
    DATASET_PATH,
    target_size=(224,224),
    batch_size=32,
    class_mode='categorical',
    subset='training',
    shuffle=True
)

validation_generator = train_datagen.flow_from_directory(
    DATASET_PATH,
    target_size=(224,224),
    batch_size=32,
    class_mode='categorical',
    subset='validation',
    shuffle=False
)

num_classes = train_generator.num_classes
```

```python

# Step 8: Build transfer learning model with ResNet50
base_model = ResNet50(weights='imagenet', include_top=False, input_shape=(224,224,3))
base_model.trainable = False

model = Sequential([
    base_model,
    GlobalAveragePooling2D(),
    Dropout(0.5),
    Dense(256, activation='relu'),
    Dropout(0.5),
    Dense(num_classes, activation='softmax')
])

model.compile(optimizer=tf.keras.optimizers.Adam(1e-4),
              loss='categorical_crossentropy',
              metrics=['accuracy'])

model.summary()
```

```python

# Step 9: Setup callbacks
earlystop = EarlyStopping(monitor='val_loss', patience=7, restore_best_weights=True)
reducelr = ReduceLROnPlateau(monitor='val_loss', factor=0.5, patience=3, min_lr=1e-7)
checkpoint = ModelCheckpoint('best_sugarcane_model.keras', monitor='val_loss', save_best_only=True)

callbacks = [earlystop, reducelr, checkpoint]
```

```python

# Step 10: Train initial model
history = model.fit(
    train_generator,
    validation_data=validation_generator,
    epochs=15,
    callbacks=callbacks
)
```

```python

# Step 11: Fine-tune deeper base model layers
base_model.trainable = True
for layer in base_model.layers[:-50]:
    layer.trainable = False

model.compile(optimizer=tf.keras.optimizers.Adam(1e-5),
              loss='categorical_crossentropy',
              metrics=['accuracy'])

history_fine = model.fit(
    train_generator,
    validation_data=validation_generator,
    epochs=10,
    callbacks=callbacks
)
```

```python

# Step 12: Evaluate performance with confusion matrix and classification report
validation_generator.reset()
pred_probs = model.predict(validation_generator)
pred_labels = np.argmax(pred_probs, axis=1)
true_labels = validation_generator.classes
class_names = list(train_generator.class_indices.keys())

cm = confusion_matrix(true_labels, pred_labels)
plt.figure(figsize=(10,8))
sns.heatmap(cm, annot=True, fmt='d', xticklabels=class_names, yticklabels=class_names, cmap='Blues')
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.title('Confusion Matrix')
plt.show()

print(classification_report(true_labels, pred_labels, target_names=class_names))
```

```python

# Step 13: Save fine-tuned model
model.save("sugarcane_resnet50_final.keras")
print("Training complete and model saved as sugarcane_resnet50_final.keras")
```

```python

# Step 14: Plot training curves
def plot_history(h1, h2=None):
    plt.figure(figsize=(14,5))
    plt.subplot(1,2,1)
    plt.plot(h1.history['accuracy'], label='Train')
    plt.plot(h1.history['val_accuracy'], label='Validation')
    if h2:
        plt.plot(h2.history['accuracy'], label='Fine-tune Train')
        plt.plot(h2.history['val_accuracy'], label='Fine-tune Validation')
    plt.title('Accuracy')
    plt.legend()

    plt.subplot(1,2,2)
    plt.plot(h1.history['loss'], label='Train')
    plt.plot(h1.history['val_loss'], label='Validation')
    if h2:
        plt.plot(h2.history['loss'], label='Fine-tune Train')
        plt.plot(h2.history['val_loss'], label='Fine-tune Validation')
    plt.title('Loss')
    plt.legend()
    plt.show()

plot_history(history, history_fine)
```

```python
import os
import numpy as np
from tensorflow.keras.models import load_model
from tensorflow.keras.preprocessing.image import load_img, img_to_array
from tensorflow.keras.applications.resnet50 import preprocess_input
from PIL import Image
import matplotlib.pyplot as plt

# Load saved model
loaded_model = load_model("sugarcane_resnet50_final.keras")

# Prediction function
def predict_image(image_path):
    img = load_img(image_path, target_size=(224, 224))
    img_array = img_to_array(img)
    img_array = preprocess_input(img_array)
    img_array = np.expand_dims(img_array, axis=0)  # Add batch dimension

    prediction = loaded_model.predict(img_array)[0]
    # Get class indices from the train_generator defined in cell C0Wbrrd2jKT7
    class_label = list(train_generator.class_indices.keys())[np.argmax(prediction)]
    confidence = np.max(prediction)
    return class_label, confidence

# Path to image
image_path = "sugarcane_data/Healthy/healthy (103).jpeg"

# Predict and print results
label, conf = predict_image(image_path)
print(f"Prediction: {label} ({conf * 100:.2f} %)")
actual_label = image_path.split(os.path.sep)[-2]
print(f"Actual: {actual_label}")

# Display image with matplotlib
img = Image.open(image_path)
plt.imshow(img)
plt.title(f"Predicted: {label} ({conf * 100:.2f} %)")
plt.axis('off')
plt.show()
```
